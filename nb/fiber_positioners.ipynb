```python
import os
import sys

import numpy as np
from matplotlib import pyplot as plt

#sys.path.append('../bin')
#import positioner

# plotting
plt.style.use(os.path.join(os.path.abspath('../bin'), 'plot_style.mplstyle'))
plots_dir = '../../plots/'
```

```python
plt.rcParams['axes.grid'] = False
```

```python
r = 12/10.4/2
circle1 = plt.Circle((0, 0), r, clip_on=False, fill=True, facecolor='C0', alpha=0.2)
c1 = plt.Circle((0, 0), 1/2, clip_on=False, fill=False, color='C0', alpha=0.5, ls='--')
circle2 = plt.Circle((1, 0), r, clip_on=False, fill=True, facecolor='C0', alpha=0.2)
c2 = plt.Circle((1, 0), 1/2, clip_on=False, fill=False, color='C0', alpha=0.5, ls='--')
circle3 = plt.Circle((1/2, np.sqrt(3)/2), r, clip_on=False, fill=True, facecolor='C0', alpha=0.2)
c3 = plt.Circle((1/2, np.sqrt(3)/2), 1/2, clip_on=False, fill=False, color='C0', alpha=0.5, ls='--')

fig, ax = plt.subplots(figsize=(3, 3))

ax.add_patch(circle1)
ax.add_patch(circle2)
ax.add_patch(circle3)
ax.add_patch(c1)
ax.add_patch(c2)
ax.add_patch(c3)

ax.hlines(np.sqrt(3)/2, xmin=0, xmax=1, color='black', lw=0.7, ls='--', alpha=0.5)

ax.scatter([0, 1, 1/2], [0, 0, np.sqrt(3)/2], s=7, zorder=12)

ax.text(0.5, 0.9, '10.4 mm', horizontalalignment='center', fontsize=12, rotation=0)

# robotic arm axis and fiber ferrule
theta = np.pi/5
ax.plot([0, r*np.cos(theta)], [0, r*np.sin(theta)], color='black', lw=0.7, ls='--', alpha=0.5)
#ax.scatter([1+r*np.cos(theta)/2], [r*np.sin(theta)/2], color='C0', s=7)
ax.plot([1, 1+r/2], [0, 0], color='black', lw=0.7, ls=':')
ax.plot([1+r/2, 1+r], [0, 0], color='C0', lw=0.7, ls=':')
ax.scatter([1+r/2], [0], color='C0', s=7, zorder=11)
ferrule = plt.Circle((1+r*np.cos(theta), r*np.sin(theta)), 1.25/10.4/2, clip_on=False, fill=True, facecolor='C0')
#ax.add_patch(ferrule)
fiber = plt.Circle((1+r*np.cos(theta), r*np.sin(theta)), 0.107/10.4/2, clip_on=False, fill=True, color='white', zorder=10)
#ax.add_patch(fiber)
ax.text(-0.3+r*np.cos(theta), -0.25+r*np.sin(theta), '6 mm', horizontalalignment='center', fontsize=12, rotation=180/5)

ax.set_aspect('equal')
ax.axis('off')
plt.savefig(os.path.join(plots_dir, 'fiber_positioners_scheme.png'), dpi=300)
plt.show()
```
